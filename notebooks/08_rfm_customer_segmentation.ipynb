# Phase E: RFM Customer Segmentation & CLTV
## Behavioral Clustering: Recency, Frequency, and Monetary Scoring
### Methodology:
- **Recency (R):** Days since last purchase (Quintiles 1 to 5, where 5 is most recent).
- **Frequency (F):** Total order count (Classified into 1-order, 2-order, 3+ order buckets).
- **Monetary (M):** Total monetary spend in R$ (Quintiles 1 to 5).
- **Segments:** Champions, Loyal Customers, New / Potential Loyalists, At Risk / Cannot Lose Them, Hibernating, Lost / Churned.


```
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

DATA_DIR = '../data/processed'
rfm = pd.read_csv(os.path.join(DATA_DIR, 'customer_rfm_segments.csv'))

# Summary by Segment
segment_summary = rfm.groupby('customer_segment').agg(
    customer_count=('customer_unique_id', 'count'),
    total_revenue=('total_spent', 'sum'),
    avg_recency=('recency_days', 'mean'),
    avg_spend=('total_spent', 'mean')
).reset_index()

total_rev = segment_summary['total_revenue'].sum()
segment_summary['revenue_pct'] = (segment_summary['total_revenue'] / total_rev) * 100
segment_summary = segment_summary.sort_values(by='total_revenue', ascending=False)
display(segment_summary)
```

### Segment Revenue vs. Customer Distribution

```
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))

colors = sns.color_palette('pastel')[0:len(segment_summary)]

# Customers Pie Chart
ax1.pie(segment_summary['customer_count'], labels=segment_summary['customer_segment'], autopct='%1.1f%%', colors=colors, startangle=140)
ax1.set_title('Customer Distribution by RFM Segment', fontsize=13, fontweight='bold')

# Revenue Bar Chart
sns.barplot(data=segment_summary, y='customer_segment', x='total_revenue', palette='Blues_r', ax=ax2)
ax2.set_title('Total Revenue Contribution by Segment (R$)', fontsize=13, fontweight='bold')
ax2.set_xlabel('Revenue (R$ Millions)', fontweight='bold')
ax2.set_ylabel('')

plt.tight_layout()
plt.show()
```
