# Phase 4 – Data Cleaning
## Objective
Identify missing values in all datasets and decide how to handle them.

```python
import pandas as pd
```

```python
customers = pd.read_csv("../data/raw/customers.csv")
orders = pd.read_csv("../data/raw/orders.csv")
order_items = pd.read_csv("../data/raw/order_items.csv")
products = pd.read_csv("../data/raw/products.csv")
payments = pd.read_csv("../data/raw/payments.csv")
reviews = pd.read_csv("../data/raw/reviews.csv")
sellers = pd.read_csv("../data/raw/sellers.csv")
geolocation = pd.read_csv("../data/raw/geolocation.csv")
category_translation = pd.read_csv("../data/raw/category_translation.csv")
```

```python
datasets = {
    "Customers": customers,
    "Orders": orders,
    "Order Items": order_items,
    "Products": products,
    "Payments": payments,
    "Reviews": reviews,
    "Sellers": sellers,
    "Geolocation": geolocation,
    "Category Translation": category_translation
}
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(f"Dataset: {name}")
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print("=" * 60)
    print(f"Dataset: {name}")

    missing = (df.isnull().sum() / len(df)) * 100

    print(missing.sort_values(ascending=False))
```

# Missing Value Analysis

- These represent orders that were not yet approved or delivered.
- Keep these values.
